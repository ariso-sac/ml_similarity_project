```python
def parse_price(val) -> float:
    if pd.isna(val):
        return np.nan
    s = str(val).replace(",", "")
    nums = re.findall(r"\d+(?:\.\d+)?", s)
    if not nums:
        return np.nan
    # If range like "$10.00 - $20.00", take midpoint
    floats = [float(n) for n in nums[:2]]
    return float(np.mean(floats))


def parse_weight_grams(val) -> float:
    if pd.isna(val):
        return np.nan
    s = str(val).lower().strip()
    m = re.search(r"(\d+(?:\.\d+)?)", s)
    if not m:
        return np.nan
    num = float(m.group(1))
    if "kg" in s or "kilogram" in s:
        return num * 1000.0
    if "pound" in s or "lb" in s:
        return num * 453.592
    if "ounce" in s or "oz" in s:
        return num * 28.3495
    return num  # assume grams if unspecified


def parse_rating(val) -> float:
    if pd.isna(val):
        return np.nan
    m = re.search(r"(\d+(?:\.\d+)?)", str(val))
    return float(m.group(1)) if m else np.nan

# def parse_color(val):
#     try:
#         colour = val.strip().split("|")
#         for x in colour:
#             x = x.lower()
#             if x in cols.keys():
#                 cols[x]+=1
#             else:
#                 cols[x]=1
#         print(colour)
#         return colour
#     except:
#         x = val.lower()
#         if x in cols.keys():
#             cols[x]+=1
#         else:
#             cols[x]=1
#         return val

def extract_colors(cell):
    # Parse stringified list into actual list
    if isinstance(cell, str):
        try:
            items = ast.literal_eval(cell)
        except (ValueError, SyntaxError):
            items = [cell]
    elif isinstance(cell, list):
        items = cell
    else:
        return []

    extracted = []
    for item in items:
        if not item or not str(item).strip():
            continue
        
        text = str(item).lower()
        # Normalize typos before matching
        for wrong, right in TYPO_MAP.items():
            text = re.sub(rf'\b{wrong}\b', right, text)
            
        # Sort colors by length descending so 'navy blue' matches before 'blue'
        for color in sorted(VALID_COLORS, key=len, reverse=True):
            if re.search(rf'\b{re.escape(color)}\b', text):
                formatted = color.title()
                if formatted not in extracted:
                    extracted.append(formatted)
                # Remove matched color so 'navy blue' doesn't also trigger 'blue'
                text = re.sub(rf'\b{re.escape(color)}\b', '', text)

    return extracted
```
